# Tweakpane Notebook Test

Test the TweakpaneServer proxy — a kernel-side tweakpane API that renders real tweakpane UIs in iframes.

```typescript
// Import the adapter (registers the bridge factory) and create a pane
import { createTweakpane } from "@/tools/tweakpaneAdapter.ts"

const params = {
  speed: 50,
  color: '#ff0055',
  name: 'hello',
  visible: true,
  position: { x: 0, y: 0 },
}

const pane = createTweakpane({ title: 'My Controls' })

// Bindings — these mirror tweakpane's addBinding API
pane.addBinding(params, 'speed', { min: 0, max: 100, step: 1 })
pane.addBinding(params, 'color')
pane.addBinding(params, 'name')
pane.addBinding(params, 'visible')
pane.addBinding(params, 'position', { x: { min: -100, max: 100 }, y: { min: -100, max: 100 } })

console.log('Pane created with bindings')
```

```typescript
// Show the pane — this opens an iframe with a real tweakpane UI
pane.show()
null
```

```typescript
// Read current values from the kernel-side bound object.
// After you change a slider in the iframe, re-run this cell to see the updated value.
console.log('Current params:', params)
```

```typescript
// Listen for changes on the kernel side
pane.on('change', (ev) => {
  console.log('change event:', ev.value)
})
```

```typescript
// Show a second view — both should stay in sync
pane.show()
null
```

```typescript
// Update a value from kernel code and push to all iframes
params.speed = 75
pane.refresh()
console.log('Set speed to 75 and refreshed')
```

```typescript
// Test folders and buttons
const pane2 = createTweakpane({ title: 'Advanced' })

const settings = { volume: 80, pan: 0, muted: false }

const folder = pane2.addFolder({ title: 'Audio' })
folder.addBinding(settings, 'volume', { min: 0, max: 100 })
folder.addBinding(settings, 'pan', { min: -1, max: 1, step: 0.01 })
folder.addBinding(settings, 'muted')

pane2.addButton({ title: 'Reset' }).on('click', () => {
  settings.volume = 80
  settings.pan = 0
  settings.muted = false
  pane2.refresh()
  console.log('Reset!')
})

pane2.show()
null
```
